```python
import serial
import time

COM_PORT = 'COM4'  # 請依據您的系統修改串口埠號
BAUD_RATES = 115200

try:
    ser = serial.Serial(COM_PORT, BAUD_RATES, timeout=1)
    print(f"已成功連接至 {COM_PORT}，等待接收反應時間...\n")
    
    ser.reset_input_buffer()

    while True:
        if ser.in_waiting > 0:
            raw_data = ser.readline().decode('utf-8', errors='ignore').strip()
            
            if raw_data == "FALSE_START":
                print("⚠️ [偷跑警告] 您在紅燈或黃燈亮起時就按下了按鈕！測試終止。")
            elif raw_data == "TIMEOUT":
                print("⏱️ [逾時通知] 綠燈亮起超過 3 秒未動作，測試終止。")
            elif raw_data.startswith("SUCCESS"):
                # 取得整數微秒並計算成秒數
                _, us_str = raw_data.split(":")
                us = int(us_str)
                seconds = us / 1e6
                ms = us / 1e3
                
                print(f"🎉 [測試成功] 反應時間：{us} μs | {ms:.2f} ms | {seconds:.6f} 秒")
            else:
                print(f"收到其他訊息：{raw_data}")
            
            print("-" * 50)

except KeyboardInterrupt:
    print("\n使用者停止程式。")
except Exception as e:
    print(f"發生錯誤: {e}")
finally:
    if 'ser' in locals() and ser.is_open:
        ser.close()
        print("串口已關閉。")
```
